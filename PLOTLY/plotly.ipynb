## 2. Types of Plots (Quick Reference)

* **Scatter Plot (`go.Scatter` / `px.scatter`):** Do **continuous (numerical)** variables ke beech relationship ya correlation dekhne ke liye use hota hai *(e.g., Batsman Average vs Strike Rate)*.
* **Line Chart (`mode='lines'`):** Time-series data ya kisi continuous sequence mein trend aur change over time dikhane ke liye use hota hai.
* **Bar Plot (`go.Bar` / `px.bar`):** Categorical aur numerical data ko compare karta hai. Iske 3 main types hain:
  1. **Simple Bar Plot:** Ek single numerical value ko alag-alag categories ke beech compare karta hai.
  2. **Grouped Bar Plot (`barmode='group'`):** Ek category ke andar multiple sub-categories ko side-by-side compare karta hai.
  3. **Stacked Bar Plot (`barmode='stack'`):** Sub-categories ko ek ke upar ek stack karke total aur individual contribution dono dikhata hai.
* **Bubble Plot:** Yeh ek Scatter Plot hi hai jisme **3rd numerical variable** ko marker ke `size` (aur color) se represent kiya jata hai.
* **Box Plot (`go.Box` / `px.box`):** Data ka statistical distribution (Min, Q1, Median, Q3, Max) aur **outliers** detect karne ke liye use hota hai.
* **Histogram (`go.Histogram` / `px.histogram`):** Ek single continuous variable ki frequency distribution (bins ke form mein) dikhata hai.
* **Distplot (`ff.create_distplot`):** Histogram, KDE (smooth density curve), aur Rug plot ko combine karke distribution ka detailed view deta hai.
* **Heatmap (`go.Heatmap` / `px.imshow`):** 2D matrix ya tabular values ko color intensity ke through represent karta hai *(e.g., Correlation Matrix)*.

```python
import numpy as np
import pandas as pd
import plotly.offline as pyo
import plotly.graph_objs as go
```

```python
match = pd.read_csv('matches.csv')
delivery = pd.read_csv('deliveries.csv')

ipl = delivery.merge(match, left_on='match_id', right_on='id')
ipl.head()
```

# Scatter plots are drawn between two continuous variables
# Problem :- We are going to draw a scatter plot between Batsman Avg(X axis) and
# Batsman Strike Rate(Y axis) of the top 50 batsman in IPL(All time)

```python
# Avg vs SR graph of Top 50 batsman(in terms of total runs)
# Fetching a new dataframe with Top 50 batsman
top50=ipl.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False).head(50).index.tolist()
new_ipl=ipl[ipl['batsman'].isin(top50)]
new_ipl
```

```python
# Calculating SR
# SR=[(number of runs scored)/(number of balls played)]*100
runs=new_ipl.groupby('batsman')['batsman_runs'].sum()
balls=new_ipl.groupby('batsman')['batsman_runs'].count()

sr=(runs/balls)*100
sr
```

```python
sr.rename('Strike_rate')
```

```python
# Calculating Avg
# Avg=(Total number of Runs)/(Number of outs)
# Calculating number of outs for top 50 batsman
out=ipl[ipl['player_dismissed'].isin(top50)]

nouts=out['player_dismissed'].value_counts()
nouts

```

```python
avg=runs/nouts

avg=avg.reset_index()
avg.rename(columns={'index':'batsman',0:'avg'},inplace=True)

avg=avg.merge(sr,on='batsman')
avg
```

# Scatter plot

```python
# Plot Scatter Plot here

trace=go.Scatter(x=avg['avg'],y=avg['batsman_runs'],mode='markers')

data=[trace]

layout=go.Layout(title='Batsman Avg vs SR', xaxis={'title':'Batsman Average'},yaxis={'title':'Batsman Strike Rate'})

fig=go.Figure(data=data,layout=layout)

pyo.plot(fig, filename='scatter')
```

```python
# Plot Scatter Plot here with batsman name

trace=go.Scatter(x=avg['avg'],y=avg['batsman_runs'],mode='markers', text=avg['batsman'], marker={'color':'#00a65a', 'size':16})

data=[trace]

layout=go.Layout(title='Batsman Avg vs SR', xaxis={'title':'Batsman Average'},yaxis={'title':'Batsman Strike Rate'})

fig=go.Figure(data=data,layout=layout)

pyo.plot(fig, filename='scatter')
```

```python
# Multiple Line Charts

def batsman_comp(*name):
    data=[]
    for i in name:
        single=ipl[ipl['batsman']==i]
        performance=single.groupby('season')['batsman_runs'].sum().reset_index()

        trace=go.Scatter(x=performance['season'],y=performance['batsman_runs'], mode='lines+markers',name=i)

        data.append(trace)

    layout=go.Layout(title='Batsman Record Comparator', xaxis={'title':'Season'},yaxis={'title':'Runs'})

    fig=go.Figure(data=data,layout=layout)

    pyo.plot(fig,filename='year_by_year')
```

# Dis plot

```python
import plotly.express as px

fig = px.histogram(
    avg,
    x='avg',
    marginal='rug',
    title='Average Distribution'
)

fig.show()
```

```python
import plotly.express as px
import plotly.graph_objects as go
import numpy as np
from scipy.stats import gaussian_kde

# Remove null values
data = avg['avg'].dropna()

# KDE calculation
kde = gaussian_kde(data)

x = np.linspace(data.min(), data.max(), 100)
y = kde(x)

# Histogram
fig = px.histogram(
    data,
    histnorm='probability density',
    title='Average Distribution'
)

# KDE Curve
fig.add_trace(
    go.Scatter(
        x=x,
        y=y,
        mode='lines',
        name='KDE'
    )
)

fig.show()
```

```python
from scipy.stats import gaussian_kde

hist_data = [avg['avg'].dropna(), avg['batsman_runs'].dropna()]
group_labels = ['Average', 'Strike']

fig = go.Figure()

for data, label in zip(hist_data, group_labels):

    # Histogram
    fig.add_trace(go.Histogram(
        x=data,
        histnorm='probability density',
        name=label,
        opacity=0.4,
        xbins=dict(
            size=2  # Bin size
        )
    ))

    # KDE
    kde = gaussian_kde(data)
    x = np.linspace(data.min(), data.max(), 200)

    fig.add_trace(go.Scatter(
        x=x,
        y=kde(x),
        mode='lines',
        name=f'{label} KDE'
    ))

fig.update_layout(
    title='Distribution Plot',
    barmode='overlay'
)

fig.show()
```

# Heatmap plot

```python
six = delivery[delivery['batsman_runs']==6]
sis = six.groupby(['batting_team', 'over'])['batsman_runs'].count().reset_index()
six
```

```python
import plotly.graph_objects as go

trace = go.Heatmap(
    x=sis['batting_team'],
    y=sis['over'],
    z=sis['batsman_runs']
)

data = [trace]

layout = go.Layout(
    title='Sixes Heatmap',
    xaxis=dict(title='Batting Team'),
    yaxis=dict(title='Over')
)

fig = go.Figure(data=data, layout=layout)

pyo.plot(fig, filename='heatmap.html')
```

```python
# Side by Side Heatmap

dots = delivery[delivery['batsman_runs'] == 0]
dots = dots.groupby(['batting_team', 'over'])['batsman_runs'].count().reset_index()

from plotly import tools

trace1 = go.Heatmap(
    x=six['batting_team'], 
    y=six['over'], 
    z=six['batsman_runs'].values.tolist()
)

trace2 = go.Heatmap(
    x=dots['batting_team'], 
    y=dots['over'], 
    z=dots['batsman_runs'].values.tolist()
)

fig = tools.make_subplots(
    rows=1, 
    cols=2, 
    subplot_titles=["6's", "0's"], 
    shared_yaxes=True
)

fig.append_trace(trace1, 1, 1)
fig.append_trace(trace2, 1, 2)

pyo.plot(fig)
```
