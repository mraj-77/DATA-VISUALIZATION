### 1. Jointplot (`sns.jointplot()`)
- Do numerical variables ke beech relationship aur unka individual distribution show karta hai.
- Center mein scatter plot aur sides par distribution plots hote hain.
- **Important Parameters:** `x`, `y`, `data`, `kind`

```python
sns.jointplot(x='total_bill', y='tip', data=tips, kind='scatter')
```



````python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
````

````python
plt.style.use('fivethirtyeight')
````

````python
tips = sns.load_dataset('tips')
tips.head()
````

````python
sns.jointplot(x='total_bill', y='tip', data=tips)
````

````python
sns.jointplot(x='total_bill', y='tip', kind='hex', data=tips)
````

````python
sns.jointplot(x='total_bill', y='tip', kind='kde', data=tips)
````

````python
sns.jointplot(x='total_bill', y='tip', kind='reg', data=tips)
````
