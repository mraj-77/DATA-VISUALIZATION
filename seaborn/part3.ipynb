## Categorical Distribution Plots

- Data ka distribution aur spread (quartiles, median) dekhne ke liye use hote hain.

### a. Boxplot
- Five-number summary (Minimum, Q1, Median, Q3, Maximum) show karta hai.
- Outliers detect karne ke liye useful hai.
- **Attributes:** `x`, `y`, `hue`, `data`, `orient`.

```python
sns.catplot(x='day', y='tip', kind='box', data=tips)
```


```

````python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
````

````python
plt.style.use('fivethirtyeight')
````

````python
tips = sns.load_dataset('tips')
tips.head()
````

````python
# Univariate
sns.boxplot(tips['tip'])
````

````python
# Univariate
sns.boxplot(tips['total_bill'])
````

````python
sns.catplot(x='day', y='total_bill', kind='box', data=tips)
````

````python
sns.catplot(x='day', y='total_bill', hue='sex', kind='box', data=tips)
````
