- Numerical data ka distribution visualize karne ke liye use hota hai.
- Histogram, KDE , ECDF and Rugplot plots create kar sakta hai.
- **Important Parameters:** `data`, `x`, `hue`, `kind`, `bins`

````python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
````

````python
plt.style.use('fivethirtyeight')
````

````python
df = sns.load_dataset('titanic')
df.head()
````

````python
df.info()
````

````python
df['age'].fillna(df['age'].mean(), inplace=True)
````

````python
df.info()
````

## displot

````python
sns.displot(df['age'])
````

````python
sns.displot(df['age'], kde=True)
````

````python
sns.displot(df['age'], kde=False, bins=10)
````

````python
sns.displot(x='age', kind='kde', data=df)
````

````python
sns.displot(x='age', kind='kde', data=df, rug=True)
````

````python
sns.rugplot(x='age', data=df)
````

````python
sns.displot(df[df['survived']==1]['age'], kde=True)
sns.displot(df[df['survived']==0]['age'], kde=True)
````

````python
sns.displot(data=df, x='age', hue='survived', kind='kde')
````
