## D. Categorical-Categorical Plots

- Do categorical variables ke beech relationship ko matrix-style mein represent karne ke liye use hote hain.

### a. Heatmap
- Matrix data ki values ko colors ki intensity se represent karta hai.
- Different values ko easily compare kar sakte hain.
- **Important Parameters:** `data`, `cmap`, `annot`, `fmt`

```python
sns.heatmap(data.corr(numeric_only=True), annot=True, cmap='coolwarm')
```



````python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
````

````python
plt.style.use('fivethirtyeight')
````

````python
df = sns.load_dataset('flights')
df.head()
````

````python
x = df.pivot_table(index='year', columns='month', values='passengers', aggfunc='sum')
x
````

````python
sns.heatmap(x)
````

````python
sns.heatmap(x, cbar=False)
````

````python
sns.heatmap(x, cbar=False, linewidths=0.5)
````

````python
sns.heatmap(x, cbar=False, linewidths=0.5, annot=True)
````

````python
sns.heatmap(x, cbar=False, linewidths=0.5, annot=True, fmt='d')
````

````python
sns.heatmap(x, cbar=False, linewidths=0.5, annot=True, fmt='d', cmap='viridis')
````
