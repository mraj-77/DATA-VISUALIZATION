### C. Categorical Estimate Plots[cite: 12]
* Statistical estimation (jaise mean ya counts) ko represent karne ke liye.
  * **a. Pointplot**[cite: 12] – Point estimates aur confidence intervals ko lines se connect karke dikhata hai.
  * **b. Barplot**[cite: 12] – Mean ya kisi aur numeric value ka estimate bar ki height ke roop mein dikhata hai.
  * **c. Countplot**[cite: 12] – Har categorical variable ki frequency/count bar graph ke through show karta hai.

## C. Categorical Estimate Plots

- Statistical estimation (Mean, Count, etc.) ko represent karne ke liye use hote hain.

### a. Pointplot
- Point estimates (jaise Mean) aur Confidence Intervals show karta hai.
- Points ko lines se connect karke categories compare karta hai.
- **Important Parameters:** `x`, `y`, `hue`, `data`

```python
sns.catplot(x='day', y='tip', kind='point', data=tips)
```

### b. Barplot
- Mean ya kisi other numeric value ka estimate bar ki height se show karta hai.
- Categories ke beech values compare karne ke liye useful hai.
- **Important Parameters:** `x`, `y`, `hue`, `data`

```python
sns.catplot(x='day', y='tip', kind='bar', data=tips)
```

### c. Countplot
- Har categorical variable ki frequency (count) show karta hai.
- Y-axis par automatically observations ka count show hota hai.
- **Important Parameters:** `x`, `hue`, `data`, `order`

```python
sns.catplot(x='day', kind='count', data=tips)
```

````python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
````

````python
plt.style.use('fivethirtyeight')
````

````python
tips = sns.load_dataset('tips')
tips.head()
````

# Bar Plot

````python
sns.catplot(x='smoker', y='total_bill', kind='bar', data=tips)
````

````python
sns.catplot(x='smoker', y='total_bill', hue='sex', kind='bar', data=tips)
````

````python
# estimator --> for 
sns.catplot(x='smoker', y='total_bill', hue='sex', estimator=np.std, kind='bar', data=tips)
````

````python
# estimator --> for 
sns.catplot(x='smoker', y='total_bill', hue='sex', estimator=np.median, kind='bar', data=tips)
````

# Count plot

````python

sns.catplot(x='sex', kind='count', data=tips)
````

````python
sns.catplot(x='sex', hue='smoker', kind='count', data=tips)
````
