### b. Clustermap
- Similar rows aur columns ko group (cluster) karta hai.
- Hierarchical clustering ke through heatmap banata hai.
- **Important Parameters:** `data`, `cmap`, `annot`, `method`

```python
sns.clustermap(data.corr(numeric_only=True), annot=True, cmap='coolwarm')
```

````python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
````

````python
plt.style.use('fivethirtyeight')
````

````python
df = sns.load_dataset('flights')
df.head()
````

````python
x = df.pivot_table(index='year', columns='month', values='passengers', aggfunc='sum')
x
````

````python
sns.clustermap(x)
````

````python
sns.clustermap(x, metric='correlation')
````

````python
sns.clustermap(x, metric='correlation', z_score=0)
````

````python
sns.clustermap(x, metric='correlation', z_score=0, annot=True)
````

````python
sns.clustermap(x, metric='correlation', z_score=0, annot=True, row_cluster=False)
````

````python
sns.clustermap(x, metric='correlation', z_score=0, annot=True, col_cluster=False)
````
