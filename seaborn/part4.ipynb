## b. Violinplot
 **b. Violinplot**[cite: 11] – Boxplot aur Kernel Density Estimation (KDE) ka combination hota hai jo distribution ka shape bhi batata hai.

- Boxplot + KDE (Kernel Density Estimation) ka combination hai.
- Data distribution ka shape aur density show karta hai.
- **Attributes:** `x`, `y`, `hue`, `data`, `split`.

```python
sns.catplot(x='day', y='tip', kind='violin', data=tips)

````python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
````

````python
plt.style.use('fivethirtyeight')
````

````python
tips = sns.load_dataset('tips')
tips.head()
````

````python
sns.violinplot(tips['total_bill'])
````

````python
sns.catplot(x='day', y='total_bill', kind='violin', data=tips)
````

````python
sns.catplot(x='day', y='total_bill', hue='sex', kind='violin', data=tips)
````

````python
sns.catplot(x='day', y='total_bill', hue='sex', kind='violin', split=True, data=tips)
````
