# Seaborn: Relational & Statistical Plots

## 1. Relational Plots Overview
* **Relational Plots** ka use multivariate analysis karne ke liye kiya jata hai (yani do ya do se zyada variables ke beech ka relation samajhne ke liye)[cite: 7].
* Main types:
  1. **Scatter Plots** – Do numerical variables ke beech joint distribution aur relationship dekhne ke liye[cite: 7, 8].
  2. **Line Plots** – Especially time-series data ke liye use hota hai[cite: 7].
  3. **FacetGrids** – Multiple graphs ko side-by-side plot karne ke liye[cite: 7].

---

## 2. Scatter Plots Deep Dive
* **What is a Scatter Plot?** Yeh statistical visualization ka ek main pillar hai jo points ki cloud/cluster ke roop mein data dikhata hai[cite: 8]. Har point dataset ke ek observation ko represent karta hai[cite: 8].
* **Use case:** Yeh dekhne ke liye ki do variables ke beech koi meaningful relationship (jaise correlation) hai ya nahi[cite: 8] (e.g., House ka area badhne se sale price par kya asar padta hai)[cite: 8].
* **Data Types:** Halanki ise categorical data ke sath bhi plot kiya ja sakta hai, lekin yeh mostly **numerical data** ke beech relationship dikhane ke liye use hota hai[cite: 9].

---

## 3. Two Ways to Draw a Scatterplot in Seaborn
Seaborn mein scatter plot banane ke do main functions hain[cite: 9]:

1. **`relplot()` [Figure-Level Function]**[cite: 9]
   * Yeh poore figure/canvas ko control karta hai aur complex multi-plot grids (jaise `col` ya `row` parameters) banane ke liye best hai.
2. **`scatterplot()` [Axes-Level Function]**[cite: 9]
   * Yeh ek specific matplotlib Axes ke andar plot banata hai. Agar aapko ek single customized plot banana ho toh yeh use hota hai.

---

## 4. Mapping Additional Variables (Semantics: Hue, Size, Style)
Complex data ko asani se samajhne ke liye Seaborn mein teen extra variables tak ko map kiya ja sakta hai[cite: 9]:
* **`hue`** – Color ke through category ya group dikhane ke liye (e.g., gender ya region ke hisaab se colors badalna).
* **`size`** – Points ke size ke through koi numerical/categorical weight dikhane ke liye.
* **`style`** – Markers/shapes ke pattern ko change karke extra variable represent karne ke liye.

---

## Important Points ⭐
* **Relational Plots** = Multivariate analysis (jaise `relplot` aur `scatterplot`)[cite: 7, 9].
* **Figure-level (`relplot`) vs Axes-level (`scatterplot`)** = Grid-based vs single-axes plotting ke liye[cite: 9].
* **Semantics (`hue`, `size`, `style`)** = 2D plot par 3 ya usse zyada variables ki information ek sath map karne ka powerful tareeqa[cite: 9].

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.style.use('fivethirtyeight')
```

```python
df = sns.load_dataset('tips')
df.head()
```

# plot scatter plot bitween total_bill and tips

```python
sns.regplot(x='total_bill', y='tip', data=df)
```

```python
sns.scatterplot(x='total_bill', y='tip', data=df)
```

```python
sns.scatterplot(x='total_bill', y='tip', hue='smoker', data=df)
```

```python
sns.scatterplot(x='total_bill', y='tip', hue='smoker', style='sex', data=df)
```

```python
sns.scatterplot(x='total_bill', y='tip', hue='smoker', style='sex', size='size', data=df)
```

```python
sns.scatterplot(x='total_bill', y='tip', hue='smoker', style='sex', size='size', sizes=(15,200), data=df)
```
