# Seaborn: Plotting with Categorical Data

## 1. Introduction & `catplot()`
* **`catplot()` [Figure-Level Function]**[cite: 10]
  * Yeh ek figure-level function hai jo categorical data ko visualize karne ke liye alag-alag types ke plots ko ek common interface deta hai (jaise `relplot` relational plots ke liye karta hai).

---

## 2. Categories of Categorical Plots

### A. Categorical Scatterplots[cite: 10]
* Data points ko spread karke categorical axis par dikhane ke liye use hota hai.
  * **a. Stripplot**[cite: 10] – Points ko category ke upar scatter karke dikhata hai taaki data density pata chale.
  * **b. Swarmplot**[cite: 10] – Stripplot ki tarah hota hai lekin isme points overlap nahi hote, balki adjust ho kar bees/swarm jaisi shape banate hain.

### B. Categorical Distribution Plots[cite: 11]
* Data ka distribution aur spread (jaise quartiles aur median) dekhne ke liye.
  * **a. Boxplot**[cite: 11] – Five-number summary (minimum, Q1, median, Q3, maximum) aur outliers dikhata hai.
  * **b. Violinplot**[cite: 11] – Boxplot aur Kernel Density Estimation (KDE) ka combination hota hai jo distribution ka shape bhi batata hai.

### C. Categorical Estimate Plots[cite: 12]
* Statistical estimation (jaise mean ya counts) ko represent karne ke liye.
  * **a. Pointplot**[cite: 12] – Point estimates aur confidence intervals ko lines se connect karke dikhata hai.
  * **b. Barplot**[cite: 12] – Mean ya kisi aur numeric value ka estimate bar ki height ke roop mein dikhata hai.
  * **c. Countplot**[cite: 12] – Har categorical variable ki frequency/count bar graph ke through show karta hai.

### D. Categorical-Categorical Plots[cite: 12]
* Do categorical variables ke beech matrix-style relationship dikhane ke liye.
  * **a. Heatmaps**[cite: 12] – Color intensity ke zariye matrix data ki values ko represent karta hai.
  * **b. Clustermaps**[cite: 12] – Similar rows aur columns ko cluster karke hierarchically organized heatmap banata hai.

### E. Plotting Side-by-Side Graphs using FacetGrids[cite: 12]
* Ek hi figure ke andar multiple subplots ko categories ya conditions ke base par side-by-side plot karne ke liye `FacetGrid` ka use hota hai[cite: 12].

## Important Points ⭐
* **`catplot`** = Categorical plots ke liye figure-level master function[cite: 10].
* **Scatter vs Distribution vs Estimate** = Raw points (strip/swarm) vs Summary shape (box/violin) vs Statistical aggregation (bar/count/point)[cite: 10, 11, 12].
* **Heatmaps/Clustermaps** = Matrix ya categorical-categorical comparison ke liye best hain[cite: 12].
