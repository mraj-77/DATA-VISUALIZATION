# Pie chart

```python
roles = ['Developer','Data Analyst','ML Engineer','AI Engineer','Web Developer','Data Scientist','Python Developer',
    'Software Engineer','Cloud Engineer','Full Stack Developer'
]
budget = [25.5, 30, 35, 40, 45, 50, 55, 65, 70, 80]
```

```python
plt.pie(budget, labels=roles)
```

```python
plt.pie(budget, labels=roles)
plt.show()
```

```python
plt.pie(budget, labels=roles, autopct='%0.1f%%')
plt.show()
```

```python
area = [1200, 1500, 1800, 2000, 2500, 3000, 3500, 4000, 4500, 5000]

budget = [25, 30, 35, 40, 45, 50, 55, 65, 70, 80]
```

```python
plt.scatter(area, budget)
plt.show()
```

```python
plt.scatter(area, budget)
plt.xlabel('area')
plt.ylabel('budget')
plt.title('area vs budget')
plt.show()
```

```python
product_id = [101, 102, 103, 104, 105, 106, 107, 108, 109, 110]
online_price = [499, 799, 1299, 1599, 1999, 2499, 2999, 3499, 3999, 4999]
offline_price = [500, 750, 1200, 1500, 2000, 2500, 3000, 3500, 4000, 5000]
chor_bazaar_price = [150, 250, 400, 300, 1200, 800, 200, 500, 100, 1500]
```

```python
plt.plot(product_id, online_price)
plt.xlabel('product')
plt.ylabel('price')
plt.title('product price comparision')
plt.show()
```

```python
plt.plot(product_id, online_price, color='red')
plt.xlabel('product')
plt.ylabel('price')
plt.title('product price comparision')
plt.show()
```

```python
plt.plot(product_id, online_price, color='red', marker='+')
plt.xlabel('product')
plt.ylabel('price')
plt.title('product price comparision')
plt.show()
```

```python
plt.plot(product_id, online_price, color='red', marker='D', markersize=10)
plt.xlabel('product')
plt.ylabel('price')
plt.title('product price comparision')
plt.show()
```

```python
plt.plot(product_id, online_price, color='red', marker='D', markersize=10)
plt.plot(product_id, offline_price, color='green', markersize=10)
plt.plot(product_id, chor_bazaar_price, color='blue', marker='+', markersize=10)
plt.xlabel('product')
plt.ylabel('price')
plt.title('product price comparision')
plt.show()
```

```python
plt.plot(product_id, online_price, color='red', marker='D', markersize=10, label='online_price')
plt.plot(product_id, offline_price, color='green', markersize=10, label='offline_price')
plt.plot(product_id, chor_bazaar_price, color='blue', marker='+', markersize=10, label='chor_bazaar_price')
plt.xlabel('product')
plt.ylabel('price')
plt.title('product price comparision')
plt.show()
```

```python
plt.plot(product_id, online_price, color='red', marker='D', markersize=10, label='online_price')
plt.plot(product_id, offline_price, color='green', markersize=10, label='offline_price')
plt.plot(product_id, chor_bazaar_price, color='blue', marker='+', markersize=10, label='chor_bazaar_price')
plt.xlabel('product')
plt.ylabel('price')
plt.title('product price comparision')
plt.legend()
plt.show()
```

```python
year = [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024]
price = [2400, 3900, 3300, 4400, 5600, 2300, 4500, 8700, 7600, 1200]
```

```python
plt.bar(year, price)
```

```python
plt.bar(year, price)
plt.xlabel('year')
plt.ylabel('price (in INR)')
plt.title('year by year comparision')
plt.show()
```

```python
plt.barh(year, price, color='red')
plt.xlabel('year')
plt.ylabel('price (in INR)')
plt.title('year by year comparision')
plt.show()
```
